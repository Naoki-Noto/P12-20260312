```python
import pandas as pd
from rdkit import Chem
from rdkit import RDLogger
RDLogger.DisableLog('rdApp.*')  # RDKitの冗長な警告を抑制
```

```python
# 既存のCSVを読み込み
df = pd.read_csv('data/100OPS_B3LYP.csv')

def canon(s: str, isomeric: bool = True):
    """失敗時は None を返す安全版カノニカライズ関数"""
    if not isinstance(s, str) or not s.strip():
        return None
    try:
        mol = Chem.MolFromSmiles(s)
        if mol is None:
            return None
        # isomeric=True で立体情報を保持したカノニカルSMILES
        return Chem.MolToSmiles(mol, canonical=True, isomericSmiles=isomeric)
    except Exception:
        return None

# 新しい列を追加（立体情報ありのカノニカルSMILES）
df['SMILES_canonical'] = df['SMILES'].apply(lambda s: canon(s, isomeric=True))

# どれくらい失敗したか軽くチェック（任意）
n_none = df['SMILES_canonical'].isna().sum()
print(f'Canonical化に失敗（None）した件数: {n_none}')

# 保存（元ファイルは残し、新しいファイル名で）
df.to_csv('data/100OPS_B3LYP_with_canon.csv', index=False)
```
