```python
from GCNcalculator import predict_properties
from T5generator import gen_smiles_from_fragments
import torch
import numpy as np
import pandas as pd
```

```python
frags = ['Nc1ccc2c(c1)C(=O)NC2=O', 'Nc1ccc2c(c1)C(=O)NC2=O', 'N#Cc1cnc2cc(O)c(N)cc2c1', 'N#Cc1cnc2cc(O)c(N)cc2c1', 'N#Cc1cnc2cc(O)c(N)cc2c1']
smiles_list = gen_smiles_from_fragments(frags, max_new_tokens=128)
print(smiles_list)
```

```python
preds = predict_properties(smiles_list)
print("SMILES:", smiles_list)
print("Prediction:", preds.squeeze())
```

```python
targetOPS = np.array([-5.9155, -2.7331, 2.5561, 0.0639, 2.4273, 0.1288])
RMSEs = np.sqrt(((preds - targetOPS)**2).mean(axis=1))
print(RMSEs)
print('Min:', np.min(RMSEs))
```

```python
rewards = 1/RMSEs
print(rewards)
```

```python
assert len(smiles_list) == preds.shape[0] == rewards.shape[0]

prop_cols = ['HOMO','LUMO','ES1','fS1','ET1','DEST']

df = pd.DataFrame(preds, columns=prop_cols)
df.insert(0, 'SMILES', smiles_list)
df['reward'] = rewards
df
```
